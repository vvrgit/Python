# **Greedy Algorithm**

image.png

```python
Machine=["A","B","C"]
Rewards_A=[3,2,4]
Rewards_B=[6,7]
Rewards_c=[5,4,6]
Avg_Rewards=[sum(Rewards_A)/len(Rewards_A),sum(Rewards_B)/len(Rewards_B),sum(Rewards_c)/len(Rewards_c)]
id=Avg_Rewards.index(max(Avg_Rewards))
print(f"Choose {Machine[id]} machine")
```

```python
# Incremental Reward Estimation using Update Formula

num_machines = int(input("Enter the number of machines: "))

Q_values = []

for machine in range(1, num_machines + 1):

    print(f"\nMachine {machine}")

    num_rewards = int(input("Enter the number of rewards: "))

    Q = 0

    for i in range(1, num_rewards + 1):

        reward = float(input(f"Enter Reward {i}: "))

        # Incremental Update Formula
        Q = Q + (reward - Q) / i

        print(f"After Reward {i}: Q = {Q:.2f}")

    Q_values.append(Q)

print("\nEstimated Reward of Each Machine")
print("-" * 40)

for i, q in enumerate(Q_values, start=1):
    print(f"Machine {i}: Q = {q:.2f}")

# Greedy Selection
best_machine = Q_values.index(max(Q_values)) + 1

print("\nGreedy Selection")
print("-" * 40)
print(f"Best Machine = Machine {best_machine}")
print(f"Estimated Reward = {max(Q_values):.2f}")
```

# **rjust(width)**

```python
text = "Python"

# Right-justify with default space padding
result_default = text.rjust(10) # Total width 10, padded with spaces
print(f"Default rjust: '{result_default}'")

# Right-justify with a custom fill character
result_custom = text.rjust(10, '*') # Total width 10, padded with asterisks
print(f"Custom rjust:  '{result_custom}'")

long_text = "Hello World"
# If the string is longer than the width, it returns the original string
result_long = long_text.rjust(5)
print(f"Long string rjust: '{result_long}'")
```

**strip(): Returns a new string with leading and trailing whitespace (spaces, tabs, newlines) removed. If chars is provided, it removes those characters from both ends.**

```python
text_with_whitespace = "   Hello, World!\n\t"

# Strip leading/trailing whitespace (default behavior)
stripped_text_default = text_with_whitespace.strip()
print(f"Default strip: '{stripped_text_default}'")

# Strip specific characters from leading/trailing ends
text_with_chars = "--abc--def--"
stripped_text_chars = text_with_chars.strip('-')
print(f"Strip specific chars: '{stripped_text_chars}'")

# Strip multiple specific characters
text_mixed_chars = "-+#Hello World#+-"
stripped_mixed = text_mixed_chars.strip('+-#')
print(f"Strip mixed chars: '{stripped_mixed}'")

# rstrip() and lstrip() for one side only
text_sides = "   Python   "
lstripped_text = text_sides.lstrip()
rstripped_text = text_sides.rstrip()
print(f"lstrip: '{lstripped_text}'")
print(f"rstrip: '{rstripped_text}'")
```

```python
positive_words = [
    "good",
    "excellent",
    "great",
    "amazing",
    "awesome",]

negative_words = [
    "bad",
    "poor",
    "worst",
    "boring",
    "slow",]
```
